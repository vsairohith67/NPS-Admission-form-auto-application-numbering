```python
from reportlab.pdfgen import canvas
from reportlab.lib.pagesizes import A4
from PyPDF2 import PdfReader, PdfWriter
import io
import copy

# =========================
# SETTINGS
# =========================

input_pdf = "Application_form.pdf"
output_pdf = "Forms_2201_to_2300_FIXED.pdf"

start_number = 2201
end_number = 2300

# Adjust if needed
x_position = 65
y_position = 697

reader = PdfReader(input_pdf)
writer = PdfWriter()

# =========================
# PROCESS
# =========================

for number in range(start_number, end_number + 1):

    # Create overlay
    packet = io.BytesIO()
    can = canvas.Canvas(packet, pagesize=A4)

    can.setFont("Courier-Bold", 18)
    can.drawString(x_position, y_position, str(number))

    can.save()
    packet.seek(0)

    overlay_pdf = PdfReader(packet)

    # VERY IMPORTANT — create fresh page
    page = copy.copy(reader.pages[0])

    page.merge_page(overlay_pdf.pages[0])

    writer.add_page(page)

# Save file
with open(output_pdf, "wb") as f:
    writer.write(f)

print("Done — no overwriting now.")
```
